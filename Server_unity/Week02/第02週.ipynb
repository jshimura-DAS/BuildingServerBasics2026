## データ型の分類
言語が提供するデータ型は、**単一の値を扱う型**と**複数の値を扱う型**に大別できる。
ここでは複数の値を扱う型に焦点を当てる。

### 1. 配列に近い型: リスト（list）
リストは**順序あり・変更可能**なコレクション。ゲームのスコア一覧、ログ一覧などに向く。

```python
# list_sample.py
scores = [1200, 980, 1430, 1100]

scores.append(1250)          # 末尾追加
top_score = max(scores)      # 最大値
avg_score = sum(scores) / len(scores)

print("scores:", scores)
print("top:", top_score)
print("average:", round(avg_score, 1))
```

#### リストの要素を1つずつ処理する（for）
`for`は、リストの要素を先頭から順に取り出して、同じ処理を繰り返す。書き方は目的によって使い分ける。

**書き方1: 要素を直接取り出す（最も基本）**

```python
# list_for1.py
scores = [1200, 980, 1430, 1100]

for s in scores:
    print("score:", s)
```

- 要素の値だけが必要なときに使う。最も短く読みやすい。

**書き方2: 添字（インデックス）で取り出す**

```python
# list_for2.py
scores = [1200, 980, 1430, 1100]

for i in range(len(scores)):
    print(i, scores[i])
```

- `range(len(scores))`は`0, 1, 2, 3`を順に作る。
- 位置番号が必要なとき、リストの要素を書き換えたいときに使う。ただし、書き方3の方が簡潔に書ける場合が多い。

**書き方3: 番号と要素を同時に取り出す（enumerate）**

```python
# list_for3.py
scores = [1200, 980, 1430, 1100]

for i, s in enumerate(scores, start=1):
    print(f"{i}人目: {s}")
```

- `enumerate`は`(番号, 要素)`の組を作る。`start=1`で番号を1から始められる。
- 「1位、2位…」のような順位表示に向く。

**書き方4: 条件付きで処理する（for + if）**

```python
# list_for4.py
scores = [1200, 980, 1430, 1100]

for s in scores:
    if s >= 1100:
        print(s, "は合格")
    else:
        print(s, "は不合格")
```

**書き方5: 内包表記（新しいリストを作る）**

```python
# list_for5.py
scores = [1200, 980, 1430, 1100]

bonus_scores = [s + 100 for s in scores]
high_scores = [s for s in scores if s >= 1100]

print(bonus_scores)
print(high_scores)
```

- `[式 for 変数 in リスト]`: 全要素に処理を行った新しいリストを作る。
- `[式 for 変数 in リスト if 条件]`: 条件に合う要素だけを集める。
- 元のリストは変更されない。

**書き方の使い分け**

| 目的 | 書き方 |
| ---- | ------ |
| 要素の値だけ使う | 書き方1 |
| 番号も使う（順位表示など） | 書き方3 |
| 添字で位置を指定して書き換える | 書き方2 |
| 条件に合う要素だけ処理する | 書き方4 |
| 加工・抽出した新しいリストを作る | 書き方5 |

#### 範囲を指定して取り出す（スライス）
リストは添字で位置を指定する。**先頭の添字は0**で、`-1`は末尾を表す。`:`を使うと、範囲をまとめて取り出せる。

```python
# list_slice.py
scores = [1200, 980, 1430, 1100, 1250, 900]
#  添字:     0     1     2     3     4    5

print(scores[0])      # 1200（先頭）
print(scores[-1])     # 900（末尾）

print(scores[1:4])    # [980, 1430, 1100]（1以上4未満）
print(scores[:3])     # [1200, 980, 1430]（先頭から3個）
print(scores[3:])     # [1100, 1250, 900]（3番目から最後まで）
print(scores[-3:])    # [1100, 1250, 900]（末尾から3個）
print(scores[::2])    # [1200, 1430, 1250]（1つおき）
print(scores[::-1])   # [900, 1250, 1100, 1430, 980, 1200]（逆順）
```

書式は`リスト[開始:終了:間隔]`。

- 開始は含み、**終了は含まない**（`[1:4]`は添字1, 2, 3）。
- 省略すると、開始は先頭、終了は末尾、間隔は1になる。
- スライスは新しいリストを返す。元のリストは変わらない。

実用例: スコア降順に並べたリストから上位3件を取り出す。

```python
# list_top3.py
scores = [1200, 980, 1430, 1100, 1250, 900]

top3 = sorted(scores, reverse=True)[:3]
print("上位3件:", top3)
```

#### 条件を使った範囲指定（抽出）
`:`による指定は**位置**での範囲指定。**値の条件**で絞り込むときは、`if`を使う。

```python
# list_filter.py
scores = [1200, 980, 1430, 1100, 1250, 900]

# 1000以上のスコアだけ抽出
passed = [s for s in scores if s >= 1000]

# 1000以上1300未満（範囲条件）
middle = [s for s in scores if 1000 <= s < 1300]

# 条件に合う要素の件数と合計
print("1000以上:", passed)
print("1000以上1300未満:", middle)
print("件数:", len(passed), "合計:", sum(passed))
```

- Pythonでは`1000 <= s < 1300`のように、範囲条件を連続して書ける。
- 位置で絞る→スライス`[:]`、値で絞る→`if`付きの内包表記（または`for`と`if`）、と使い分ける。

### 2. タプル（tuple）
タプルは**順序あり・変更不可**なコレクション。座標、固定設定値など「変えてはいけない値」の管理に向く。

```python
# tuple_sample.py
spawn_point = (10, 20, 5)  # x, y, z

x, y, z = spawn_point
distance_hint = x + y + z

print("spawn:", spawn_point)
print("x:", x, "y:", y, "z:", z)
print("distance_hint:", distance_hint)
```

---

### 3. 辞書（dict）
辞書は**キーと値**の対応で管理する型。JSONに近く、API開発で頻出。

```python
# dict_sample.py
player = {
    "name": "PlayerA",
    "level": 7,
    "hp": 85
}

player["hp"] -= 20
player["items"] = ["Potion", "Key"]

print("name:", player["name"])
print("hp:", player["hp"])
print("items:", player["items"])
```

### 4.集合型(set)
集合型は**順序なし・重複なし**のコレクション。ユニークな値の管理に向く。
重複したものを除外したい場合に便利で、リストやタプルのように順序を気にしない場合に使う。

```python
# set_sample.py
unique_items = {"Potion", "Key", "Sword", "Shield"}
unique_items.add("Arrow")
print("unique_items:", unique_items)
unique_items.add("Potion")
print("unique_items:", unique_items)
```

### 5. データ型の使い分け
それぞれの型で特徴を整理すると以下の通り。
| データ型 | 特徴        | 用途例           |
| ---- | --------- | ------------- |
| リスト  | 順序あり・変更可能 | スコア一覧、ログ一覧    |
| タプル  | 順序あり・変更不可 | 座標、固定設定値      |
| 辞書   | キーと値の対応   | JSON、APIレスポンス |
| 集合型  | 順序なし・重複なし | ユニークな値の管理     |

## データ型の組み合わせ
これらの型は、単独で使うよりも組み合わせて使うことで、より複雑なデータ構造を表現できる。

### 1. リスト + 辞書の組み合わせ実用例（意味のある処理）
ランキングデータを作り、スコア順に並べ替えて上位3件を表示する。

```python
# ranking_sample.py
players = [
    {"name": "A", "score": 950},
    {"name": "B", "score": 1330},
    {"name": "C", "score": 1210},
    {"name": "D", "score": 1440},
]

sorted_players = sorted(players, key=lambda p: p["score"], reverse=True)

print("=== TOP 3 ===")
for i, p in enumerate(sorted_players[:3], start=1):
    print(f"{i}位 {p['name']} : {p['score']}")
```

### 基本利用例

```python
# pandas_sample.py
import pandas as pd

data = [
    {"name": "A", "score": 950},
    {"name": "B", "score": 1330},
    {"name": "C", "score": 1210},
    {"name": "D", "score": 1440},
]

df = pd.DataFrame(data)

print("平均スコア:", df["score"].mean())
print("1200以上:\n", df[df["score"] >= 1200])
```

### 組み込み型との違い（要点）
- 組み込み型（list/dict）: 軽量で基礎学習に最適。小規模データやロジック学習向き。
- Pandas: 集計・抽出・欠損値処理・CSV入出力が強力。中〜大規模データで有利。
- 使い分け: まず組み込み型で構造を理解し、データ量や分析要件が増えたらPandasを導入する。
