## 1. リスト（list）
順序あり・変更可能。スコア一覧などに向く。

```python
scores = [1200, 980, 1430, 1100]
scores.append(1250)

top_score = max(scores)
avg_score = sum(scores) / len(scores)

print('scores:', scores)
print('top:', top_score)
print('average:', round(avg_score, 1))
```

## 2. タプル（tuple）
順序あり・変更不可。固定値（座標など）に向く。

```python
spawn_point = (10, 20, 5)  # x, y, z
x, y, z = spawn_point
distance_hint = x + y + z

print('spawn:', spawn_point)
print('x:', x, 'y:', y, 'z:', z)
print('distance_hint:', distance_hint)
```

## 3. 辞書（dict）
キーと値の対応で管理する。JSONに近い。

```python
player = {
    'name': 'PlayerA',
    'level': 7,
    'hp': 85
}

player['hp'] -= 20
player['items'] = ['Potion', 'Key']

print('name:', player['name'])
print('hp:', player['hp'])
print('items:', player['items'])
```

## 4. 集合（set）
順序なし・重複なし。ユニーク値管理に向く。

```python
unique_items = {'Potion', 'Key', 'Sword', 'Shield'}
unique_items.add('Arrow')
print('1回目:', unique_items)

unique_items.add('Potion')  # 既存要素なので増えない
print('2回目:', unique_items)
```

## 5. 型の組み合わせ実用例（ランキング）
リスト + 辞書を使って上位3件を表示する。

```python
players = [
    {'name': 'A', 'score': 950},
    {'name': 'B', 'score': 1330},
    {'name': 'C', 'score': 1210},
    {'name': 'D', 'score': 1440},
]

sorted_players = sorted(players, key=lambda p: p['score'], reverse=True)

print('=== TOP 3 ===')
for i, p in enumerate(sorted_players[:3], start=1):
    print(f"{i}位 {p['name']} : {p['score']}")
```

## 6. 外部ライブラリ Pandas
標準型でも処理はできるが、集計・抽出・CSV処理はPandasが得意。

```python
# Colabでは通常pandasは利用可能。必要なら以下を実行。
# !pip install pandas
```

```python
import pandas as pd

data = [
    {'name': 'A', 'score': 950},
    {'name': 'B', 'score': 1330},
    {'name': 'C', 'score': 1210},
    {'name': 'D', 'score': 1440},
]

df = pd.DataFrame(data)
print('平均スコア:', df['score'].mean())
print('1200以上:')
display(df[df['score'] >= 1200])
```

### 組み込み型との違い（要点）
- list/dict: 軽量。基礎学習、小規模データ向き
- Pandas: 集計・抽出・欠損値処理・CSV入出力が強力
- 使い分け: まず組み込み型で理解し、要件拡大時にPandas導入
